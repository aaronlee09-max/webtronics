## 단계 1 문제 정의하기

손으로 쓴 숫자 분류 모델을 구현해 보자.

## 단계 2 데이터 수집 및 전처리하기

### (1) 데이터 수집하기

MNIST 데이터셋은 케라스 라이브러리에 저장되어 있어 바로 사용할 수 있다. 데이터 파일은 숫자가 저장된 타깃(label)과 특징(픽셀값이 저장된 784개의 컬럼)으로 구성되어 있다. 훈련 데이터 파일은 60,000개, 테스트 데이터 파일은 10,000개이다.

### (2) 데이터 전처리하기

① 케라스 라이브러리에 저장된 MNIST 데이터셋을 불러와 훈련 데이터와 테스트 데이터로 분할한다.

```python
from tensorflow import keras

mnist = keras.datasets.mnist
(X_train, y_train), (X_test, y_test) = mnist.load_data()
print(X_train.shape, y_train.shape, X_test.shape, y_test.shape)
```

**❓ 확인하기**

- 훈련 데이터와 테스트 데이터는 각각 몇 개인가?  → 훈련 **60000**개 · 테스트 **10000**개
- `X_train`과 `X_test`의 한 이미지가 `(28, 28)`로 표시되는 것은 무엇을 뜻하는가?

  → **숫자 한 장이 세로 28픽셀, 가로 28픽셀이라는 뜻이다.**
- 데이터가 제대로 분할되었으므로, 다음에는 무엇을 확인해 보면 좋을까?

  → **테스트 이미지 한 장을 그림으로 확인한다.**


mnist 데이터셋은 28×28 크기의 픽셀값으로, 훈련 데이터는 60,000개, 테스트 데이터는 10,000개로 구분되어 있다.

② 테스트 데이터 0번 특징값(X_test[0])을 28×28 형태로 변경하여 보여 준다.

```python
import matplotlib.pyplot as plt

image = X_test[0].reshape(28, 28)
plt.imshow(image, cmap='gray')
plt.title(y_test[0])
plt.show()
```

**❓ 확인하기**

- 화면에 표시된 0번 테스트 데이터는 어떤 숫자인가?  → **7** (`y_test[0]`)
- 이미지 제목과 손글씨 모양이 서로 일치하는가?  → **일치한다.**
- 픽셀값의 범위가 크면 학습이 잘 이루어지지 않을 수 있다. 다음 단계에서 어떤 처리가 필요한가?

  → **0~1 사이로 정규화한다.**


③ MNIST 데이터는 0에서 255 사이의 픽셀값을 가진 28×28 크기의 데이터이다. 이 데이터를 그대로 사용하면 픽셀값의 범위가 너무 커서 학습이 잘 이루어지지 않을 수 있으므로 픽셀값의 범위를 0에서 1 사이의 값으로 정규화를 수행한다.

```python
X_train = X_train / 255.0
X_test = X_test / 255.0
```

## 단계 3 신경망 모델 생성하기

### (1) 신경망 모델 설계하기

① 케라스 라이브러리로 신경망 모델을 만드는 데 필요한 명령어를 불러온다.

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten
```

- Sequential: 층을 순차적으로 연결하는 모델 구조이다.
- Dense: 완전 연결 계층으로 설계한다.
- Flatten: 다차원 배열을 1차원 배열로 만든다.

② 신경망 모델 설계하기: 입력층 노드 개수는 데이터의 픽셀 개수(28×28=784개)로 Flatten으로 2차원 배열을 1차원으로 변환한다. 또한 출력층에는 0~9까지의 숫자가 확률값으로 출력되므로 10개의 노드가 필요하다.

```python
model = Sequential()
model.add(Flatten(input_shape=(28, 28)))
model.add(Dense(128, activation='relu'))
```

```python
model.add(Dense(64, activation='relu'))
model.add(Dense(10, activation='softmax'))
```

### (2) 신경망 모델 환경 설정하기

신경망 모델의 손실함수(loss), 최적화 함수(optimizer), 평가 지표(metrics)를 설정한다. mnist 데이터 분류 문제 손실함수는 ‘sparse_categorical_crossentropy’, 최적화 함수는 ‘adam’을 사용한다. 평가 지표는 정확도(accuracy)를 사용한다.

```python
model.compile(
    loss='sparse_categorical_crossentropy',
    optimizer='adam',
    metrics=['accuracy']
)
```

- 손실함수: sparse_categorical_crossentropy
- 최적화 함수: adam
- 평가 지표: accuracy

### (3) 신경망 모델 학습하기

60,000개의 훈련 데이터(특징 X_train, 타깃 y_train)로 총 10번을 반복 학습하여 신경망 모델을 학습한다.

```python
model.fit(X_train, y_train, epochs=10)
```

**❓ 확인하기**

- 반복 학습이 진행될수록 손실(loss)은 어떻게 변하는가?  → **줄어든다.** (교과서 예: 0.2387 → 0.0198)
- 정확도(accuracy)는 어떻게 변하는가?  → **높아진다.** (교과서 예: 0.9284 → 0.9934)
- 훈련 정확도만으로 새로운 숫자도 잘 분류한다고 판단해도 될까?

  → **안 된다. 테스트 데이터로 다시 평가해야 한다.**


분류 모델을 학습시킨 결과, 모델의 손실(loss)은 0.2387에서 0.0198으로 줄어들고, 정확도(accuracy)는 0.9284에서 0.9934로 향상되었다. 단, 신경망 모델은 훈련할 때마다 손실함수와 정확도의 값이 다를 수 있다.

## 단계 4 신경망 모델 평가 및 예측하기

① 10,000개의 테스트 데이터(X_test, y_test)를 이용해 생성된 모델의 성능을 평가한다.

```python
model.evaluate(X_test, y_test)
```

**❓ 확인하기**

- 테스트 데이터의 손실(loss)과 정확도(accuracy)는 각각 얼마인가?  → 교과서 실행 예: 손실 **0.0955** · 정확도 **0.9784**
- 훈련 정확도보다 테스트 정확도가 조금 낮은 것은 무엇을 뜻하는가?

  → **배우지 않은 데이터에서는 성능이 조금 떨어질 수 있다.**
- 모델의 전체 성능을 확인했으므로, 다음에는 무엇을 확인해 보면 좋을까?

  → **한 장이 어떤 숫자로 예측되는지 본다.**


테스트 데이터로 모델을 성능을 평가한 결과, 손실(loss)은 0.0955, 정확도(accuracy)는 0.9784이다. 단, 신경망 모델의 손실함수와 정확도의 값은 다르다.

② 테스트 데이터(X_test)를 예측한다.

```python
predictions = model.predict(X_test)
print(predictions[0])
```

**❓ 확인하기**

- 0번 테스트 데이터의 예측 결과에 확률값이 몇 개 출력되었는가?  → **10**개
- 가장 큰 확률값은 몇 번째 인덱스에 있는가?  → **7번**
- 가장 큰 확률값의 위치를 한 번에 찾으려면 다음 단계에서 어떤 함수를 사용하면 되는가?  → **np.argmax()**


예측 결과, 출력층 10개의 노드에 출력값이 저장되어 있다. 소프트맥스(softmax) 함수는 출력층 노드값의 총합이 1이 되도록 조정하는 함수로, 숫자 0부터 9까지일 확률값을 배열로 저장한다. 배열의 7번 인덱스 확률값이 9.9999851e-01로 가장 크므로 7로 예측할 수 있다.

```python
import numpy as np

print(np.argmax(predictions[0]))
```

**❓ 확인하기**

- `np.argmax(predictions[0])`의 결과는 얼마인가?  → **7**
- 앞에서 확인한 실제 레이블 `y_test[0]`과 예측 결과가 일치하는가?  → **일치한다.**
- 학습한 모델을 다른 숫자 이미지의 분류에도 사용하려면 무엇을 해야 하는가?  → **모델을 파일로 저장한다.**


넘파이(numpy)의 argmax() 함수를 사용하면 배열의 가장 큰 인덱스값을 찾아 준다. 배열의 가장 큰 값의 인덱스는 7이므로 7로 인식할 수 있다.

③ 생성한 신경망 모델을 저장한 후, mnist_model.h5 파일을 내 컴퓨터에 다운로드한다.

```python
model.save("mnist_model.h5")
```
