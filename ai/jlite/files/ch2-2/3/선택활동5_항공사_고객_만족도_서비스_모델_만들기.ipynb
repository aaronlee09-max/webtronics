### 1. 특징 중요도 확인하기

- **속성 설명:** 학습한 결정트리의 `dt.feature_importances_`는 각 특징이 분류 기준을 만드는 데 얼마나 기여했는지를 0~1 사이의 값으로 나타낸다. 값이 클수록 분류에 많이 사용되었다는 뜻이다.
- **이번 활동:** `dt.feature_importances_`를 확인한다.

### 2. 특징 이름과 중요도 묶기

- **문법 설명:** `pd.Series(데이터, index=이름)`은 데이터에 이름표(인덱스)를 붙여 하나의 표(시리즈)로 만든다.
- **이번 활동:** `pd.Series(dt.feature_importances_, index=X.columns)`로 특징 이름과 중요도를 묶어 `importances`에 저장한다.

### 3. 큰 값부터 정렬해 상위 5개 뽑기

- **문법 설명:** `.sort_values(ascending=False)`는 값을 큰 순서대로 정렬한다. `.head(5)`는 앞에서 5개만 선택한다.
- **이번 활동:** `importances.sort_values(ascending=False).head(5)`로 중요도가 큰 상위 5개 서비스를 뽑는다.
