**💡 새 API: 혼동 행렬**

### 1. 혼동 행렬 계산하기

- **API 설명:** `confusion_matrix(실젯값, 예측값)`은 실제 클래스별로 모델이 무엇으로 예측했는지 표(행렬)로 만든다. 행은 실제 클래스, 열은 예측한 클래스를 나타낸다.
- **사용 형식:** `confusion_matrix(y_test, 예측값)`
- **이번 활동:** 로지스틱 회귀의 예측값(`lr.predict(X_test)`)과 `y_test`로 혼동 행렬을 구해 `conf`에 저장한다.

### 2. 히트맵으로 시각화하기

- **API 설명:** `sns.heatmap(데이터, annot=True, fmt='d')`는 표의 값을 색과 함께 셀 안에 숫자로 표시한다. `fmt='d'`는 숫자를 정수로 표시한다.
- **이번 활동:** `sns.heatmap(conf, annot=True, fmt='d')`를 실행해 혼동 행렬을 히트맵으로 그린다.
