### 군집별 특징 해석하기

**💡 새 API: `groupby()`**

### 1. 군집별로 묶어 평균 구하기

- **API 설명:** `df.groupby('기준 열')`은 기준 열의 같은 값을 가진 행끼리 묶는다. 뒤에 `[[특징 열들]].mean()`을 붙이면 묶인 그룹별로 해당 특징의 평균을 계산한다.
- **사용 형식:** `df.groupby('cluster')[['특징1', '특징2']].mean()`
- **이번 활동:** `df.groupby('cluster')[['sugars(g)', 'caffeine(mg)']].mean()`을 계산해 `profile`에 저장한다.

- **문법 설명:** `.sum(axis=1)`은 각 행(군집)의 값을 가로로 더한다. `.idxmin()`은 그 값이 가장 작은 행의 이름(인덱스)을 반환한다.

- **이번 활동:** `profile.sum(axis=1).idxmin()`으로 설탕과 카페인 평균의 합이 가장 작은 군집 번호를 찾아 `low_cluster`에 저장한다.
