```python
import pandas as pd
df = pd.read_csv('caffe_menu.csv')
data = df[['sugars(g)', 'caffeine(mg)']]
print(df.info())
print(data.isnull().sum())
```

```python
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
data_scaled = scaler.fit_transform(data)
```

```python
from sklearn.cluster import KMeans
model = KMeans(n_clusters=4, random_state=42, n_init=10)
model.fit(data_scaled)
df['cluster'] = model.labels_
```

```python
profile = df.groupby('cluster')[['sugars(g)', 'caffeine(mg)']].mean()
print(profile)
low_cluster = (profile['sugars(g)'] + profile['caffeine(mg)']).idxmin()
print('low_cluster:', low_cluster)
```

```python
rec = df[df['cluster'] == low_cluster].sort_values(['sugars(g)', 'caffeine(mg)'])
rec[['name', 'kind', 'sugars(g)', 'caffeine(mg)']].head(10)
```

```python
max_sugar = 10
max_caffeine = 50
mine = df[(df['sugars(g)'] <= max_sugar) & (df['caffeine(mg)'] <= max_caffeine)]
mine.sort_values(['caffeine(mg)', 'sugars(g)'])[['name', 'kind', 'sugars(g)', 'caffeine(mg)']].head(10)
```

```python
from sklearn.metrics import silhouette_score
silhouette_score(data_scaled, model.labels_)
```
