# 선택 활동 2: 다중 선형 회귀로 허리둘레를 예측해 보기

목둘레(`neck`), 몸무게(`weight`), 엉덩이둘레(`heap`)를 이용해 허리둘레(`waist`)를 예측하는 다중 선형 회귀 모델을 만들어 보자.

## 단계 0: 준비 (라이브러리 설치)

```python
import importlib, sys, subprocess

packages = [
    ('pandas', 'pandas'),
    ('numpy', 'numpy'),
    ('sklearn', 'scikit-learn'),
]

for module_name, pip_name in packages:
    try:
        importlib.import_module(module_name)
    except ModuleNotFoundError:
        subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', pip_name])

from sklearn import set_config
set_config(display="text")

print('라이브러리 준비 완료')
```

## 단계 2: 데이터 수집 및 전처리하기

실습 2에서 사용한 `jeans.csv` 파일을 업로드한다.

```python
# pandas 라이브러리 가져오기
import pandas as pd
# jeans.csv를 불러와 데이터 프레임(df) 생성하기
df = pd.read_csv('jeans.csv')
# 데이터 프레임 5행 보여 주기
df.head()
```

### ① 독립 변수와 종속 변수 선정하기


```python
# neck, weight, heap 열을 독립 변수 X로 선정하기
X = df[['neck', 'weight', 'heap']]
# waist 열을 종속 변수 y로 선정하기
y = df['waist']
```

### ② 훈련 데이터와 테스트 데이터 분할하기


```python
# train_test_split 가져오기
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
print(X_train.shape, X_test.shape, y_train.shape, y_test.shape)
```

## 단계 3: 모델 생성하기


```python
# LinearRegression 가져오기
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(X_train, y_train)
print(model.coef_, model.intercept_)
```

## 단계 4: 모델 평가하기


```python
# X_test와 y_test를 이용해 모델의 결정계수 출력하기
print(model.score(X_test, y_test))
predictions = model.predict(X_test)
print(predictions[:5])
print(y_test[:5].to_numpy())
```
