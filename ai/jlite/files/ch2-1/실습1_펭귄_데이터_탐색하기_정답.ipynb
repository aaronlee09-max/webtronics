```python
import pandas as pd
df = pd.read_csv('penguins_size.csv')
df.head()
```

```python
df.describe()
```

```python
import seaborn as sns
sns.countplot(x=df['species'])
```

```python
df.isnull().sum()
```

```python
df.dropna(axis=0, inplace=True)
print(len(df))
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(6, 2))
sns.boxplot(data=df, x='culmen_length_mm')
plt.show()
```

```python
df['species'].unique()
```

```python
df['sex'].unique()
```

```python
df[df['sex'] == '.']
```

```python
df = df[df['sex'] != '.']
print(len(df))
```

```python
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
num = ['culmen_length_mm', 'culmen_depth_mm', 'flipper_length_mm', 'body_mass_g']
df[num] = scaler.fit_transform(df[num])
df[num].head()
```

```python
sns.scatterplot(data=df, x='culmen_length_mm', y='culmen_depth_mm', hue='species')
```

```python
sns.scatterplot(data=df, x='culmen_depth_mm', y='flipper_length_mm', hue='species')
```

```python
sns.regplot(data=df, x='culmen_length_mm', y='body_mass_g')
```

```python
sns.heatmap(df[num].corr(), annot=True)
```
