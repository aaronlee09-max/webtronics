```python
# 라이브러리는 JupyterLite 커널에 미리 포함되어 있습니다.
# (Colab에서는 아래 설치 코드가 실행됩니다.)
import sys

if sys.platform != "emscripten":
    import importlib, subprocess
    for module_name, pip_name in [
        ("pandas", "pandas"),
        ("numpy", "numpy"),
        ("matplotlib", "matplotlib"),
        ("seaborn", "seaborn"),
        ("sklearn", "scikit-learn"),
    ]:
        try:
            importlib.import_module(module_name)
        except ModuleNotFoundError:
            subprocess.run([sys.executable, "-m", "pip", "install", "-q", pip_name], check=False)

print("라이브러리 준비 완료")
```

```python
import pandas as pd
df = pd.read_csv('earthquake.csv', encoding='utf-8')
df.head()
```

```python
df.isnull().sum()
```

```python
df.dropna(axis=0, inplace=True)
```

```python
df['year'] = pd.to_datetime(df['시간']).dt.year
df['year'].value_counts().sort_index()
```

```python
import seaborn as sns
sns.scatterplot(data=df, x='year', y='규모')
```

```python
df['지역']=df['위치'].str[:2]
df.head()
```

```python
import matplotlib.pyplot as plt
counts = df['지역'].value_counts()
plt.figure(figsize=(8, 8))
plt.pie(counts, labels=counts.index, autopct='%.1f%%')
plt.show()
```
