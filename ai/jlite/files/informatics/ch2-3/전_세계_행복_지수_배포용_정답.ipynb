# 전 세계 행복 지수 데이터 분석 정답


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False
```

```python
df = pd.read_csv('world_happiness_report_2021.csv')
print(df.shape)
df.head()
```

```python
df.info()
```

```python
happiness = df.iloc[:,[0, 1, 2, 6, 7, 8, 9, 10, 11]]
happiness.describe()
```

```python
happiness_sorted = happiness.sort_values(by='Ladder_score', ascending=False)
happiness_sorted.loc[happiness_sorted['Country_name'] == 'South Korea', ['Country_name', 'Ladder_score']]
```

```python
top10 = happiness_sorted.iloc[0:10]
plt.figure(figsize=(10, 6))
plt.barh(top10['Country_name'], top10['Ladder_score'])
plt.title('Top 10 Happiest Countries (2021)')
plt.show()
```

```python
bottom10 = happiness_sorted.tail(10)
plt.figure(figsize=(10, 6))
plt.barh(bottom10['Country_name'], bottom10['Ladder_score'])
plt.title('Bottom 10 Happiest Countries (2021)')
plt.show()
```

```python
plt.figure(figsize=(12, 8))
sns.boxplot(x='Ladder_score', y='Regional_indicator', data=happiness, orient='h')
plt.show()
```

```python
corr = happiness.corr(numeric_only=True)
corr2 = corr['Ladder_score'].sort_values(ascending=False)
plt.figure(figsize=(10, 6))
plt.bar(corr2.index, corr2.values)
plt.xticks(rotation=90)
plt.show()
```
