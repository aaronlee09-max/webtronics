```python
# CSV 자동 준비 (Colab · JupyterLite · 로컬 공통)
# 파일이 없으면 GitHub raw에서 받아 cwd에 저장 → pd.read_csv("파일명") 그대로 동작
import sys
from pathlib import Path

def ensure_csv(filename, repo_path):
    candidates = [
        Path(filename),
        Path(repo_path),
        Path("files") / repo_path,
        Path("/files") / repo_path,
        Path(repo_path).name,
    ]
    for c in candidates:
        try:
            if c.is_file():
                if not Path(filename).exists():
                    Path(filename).write_bytes(c.read_bytes())
                return filename
        except Exception:
            pass
    url = f"https://raw.githubusercontent.com/aaronlee09-max/informatics/main/{repo_path}"
    try:
        if sys.platform == "emscripten":
            from pyodide.http import open_url
            Path(filename).write_text(open_url(url).read(), encoding="utf-8")
        else:
            import urllib.request
            urllib.request.urlretrieve(url, filename)
        print("CSV 준비:", filename)
        return filename
    except Exception as e:
        print("CSV 준비 실패 → URL 직접 사용:", url, e)
        return url

ensure_csv('navermusic_genre.csv', 'informatics/ch2-3/navermusic_genre.csv')

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False
```

```python
df = pd.read_csv('world_happiness_report_2021.csv')
print(df.shape)
df.head()
```

```python
df.info()
```

```python
happiness = df.iloc[:,[0, 1, 2, 6, 7, 8, 9, 10, 11]]
happiness.describe()
```

```python
happiness_sorted = happiness.sort_values(by='Ladder_score', ascending=False)
happiness_sorted.loc[happiness_sorted['Country_name'] == 'South Korea', ['Country_name', 'Ladder_score']]
```

```python
top10 = happiness_sorted.iloc[0:10]
plt.figure(figsize=(10, 6))
plt.barh(top10['Country_name'], top10['Ladder_score'])
plt.title('Top 10 Happiest Countries (2021)')
plt.show()
```

```python
bottom10 = happiness_sorted.tail(10)
plt.figure(figsize=(10, 6))
plt.barh(bottom10['Country_name'], bottom10['Ladder_score'])
plt.title('Bottom 10 Happiest Countries (2021)')
plt.show()
```

```python
plt.figure(figsize=(12, 8))
sns.boxplot(x='Ladder_score', y='Regional_indicator', data=happiness, orient='h')
plt.show()
```

```python
corr = happiness.corr(numeric_only=True)
corr2 = corr['Ladder_score'].sort_values(ascending=False)
plt.figure(figsize=(10, 6))
plt.bar(corr2.index, corr2.values)
plt.xticks(rotation=90)
plt.show()
```
