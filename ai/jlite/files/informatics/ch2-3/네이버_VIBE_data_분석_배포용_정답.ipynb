## 네이버 VIBE data 분석


```python
import pandas as pd
import matplotlib.pyplot as plt
plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False
```

```python
df = pd.read_csv('navermusic_age.csv')
df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
sorted = df.sort_values(by='playnum', ascending=False)
plt.figure(figsize=(12,5))
plt.bar(sorted['title'][:10], sorted['playnum'][:10])
plt.xticks(rotation=45, ha='right')
plt.title('2019년 가장 많이 재생된 음악 Top 10')
plt.tight_layout()
plt.show()
```

```python
df['10대 재생수'] = df['playnum'] * df['10']
df.sort_values(by='10대 재생수', ascending=False).head(10)
```

```python
by_artist = df.groupby('artist')['10대 재생수'].sum()
by_artist_sorted = by_artist.sort_values(ascending=False).head(10)
plt.bar(by_artist_sorted.index, by_artist_sorted.values)
plt.xticks(rotation=45, ha='right')
plt.title('2019년 10대가 가장 많이 들은 가수 TOP10')
plt.tight_layout()
plt.show()
```

```python
genre_df = pd.read_csv('navermusic_genre.csv')
genre_df
```

```python
genres =['ballad', 'dance', 'hiphop', 'pop', 'rock', 'carol']
plt.figure(figsize=(12,6))
for genre in genres:
    plt.plot(genre_df['month'], genre_df[genre], marker='o', label=genre)
plt.xticks(rotation=45, ha='right')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12,5))
plt.plot(genre_df['month'], genre_df['carol'], label='캐롤', marker='o')
plt.xticks(rotation=45, ha='right')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12,5))
plt.plot(genre_df['month'], genre_df['ballad'], label='발라드', marker='o')
plt.xticks(rotation=45, ha='right')
plt.legend()
plt.tight_layout()
plt.show()
```
