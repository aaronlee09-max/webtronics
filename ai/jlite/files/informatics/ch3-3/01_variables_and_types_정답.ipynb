## 1️⃣ 변수

**변수**는 프로그래밍에서 값을 저장하는 **기억 공간**이다. 등호(`=`)를 이용하여 변수에 값을 저장하고, 저장된 값을 변경할 수 있다.

> **형식** `변수명 = 변수에 저장할 값(상수, 변수, 수식 등)`

**변수명 지정 규칙**

- 알파벳 대소문자, 한글, 언더 바(`_`), 숫자만을 사용할 수 있다.
- `if`, `for` 등과 같은 파이썬 예약어는 변수명으로 사용할 수 없다.
- 알파벳 대소문자는 구별해서 사용한다.
- 첫 글자에 숫자가 올 수 없다.

```python
# 예제 5. 변수에 값을 저장하고 값을 변경해 보자.
# a에 정수 10을 저장하기
a = 10
# b, c에 실수 2.5와 문자 'A'를 순서대로 저장하기
b , c = 2.5, 'A'
print(a, b, c)
# a에 저장된 값에 2를 곱하기
a = a * 2
print(a, b, c)
```

## 2️⃣ 자료형

**자료형**은 변수에 저장될 데이터가 어떤 유형인지 구분하는 것으로 종류에는 수치형, 불형, 문자열 등이 있다.

| 자료형 | 설명 | 예 |
|--------|------|-----|
| 정수형 `int` | 정수를 나타냄 | `3`, `-7` |
| 실수형 `float` | 소수점이 있는 수 | `1.5` |
| 불형 `bool` | 참(`True`)과 거짓(`False`) | `True` |
| 문자열 `str` | 큰따옴표나 작은따옴표로 감싼 문자 | `"apple"` |

> `type( )` : 자료의 종류를 반환한다.

```python
# 예제 6. 변수에 다양한 값을 저장하고 자료형을 출력해 보자.
a = 3
print(type(a))
b = 3.14
print(type(b))
c = 'Hello'
print(type(c))
d = True
print(type(d))
```

### 문자열과 인덱스

**인덱스(index)** : 저장된 값(요소)의 위치 데이터로 0부터 순서대로 1씩 증가한다.

```python
# 예제 7. 문자열을 다양한 형태로 출력해 보자.
a = 'HAPPY BIRTHDAY'
print(a[0], a[6], a[-1])
```

### 🔥 도전

문자열 `a`에서 **HBD**만 출력해 보자.

```python
# 도전 ✏️ 문자열 a에서 HBD만 출력해 보자.
a = "HAPPY BIRTHDAY!"
print(a[0] + a[6] + a[8])
```

## 3️⃣ 연산자

연산자의 종류에는 산술, 비교, 논리 연산자 등이 있고, 우선순위는 **산술 > 비교 > 논리** 이다.

```python
# 예제 8. 산술 연산자를 이용하여 간단한 수식을 계산해 보자.
a = 25
b = 12
print(a + b, a - b, a * b, a / b, a // b, a % b, a ** 2)
```

### ⑵ 비교 연산자

두 개 이상의 값을 비교하여 `True`(참) 또는 `False`(거짓)를 판단하는 연산자이다.

```python
# 예제 9. 비교 연산자를 이용하여 값의 크기를 비교해 보자.
n = 10
print(n > 5, n < 5, n == 10, n != 7)
```

### ⑶ 논리 연산자

논리식의 참과 거짓을 판단하는 연산자이다.

```python
# 예제 10. 논리곱(and), 논리합(or) 연산자를 사용하여 참, 거짓을 판단해 보자.
id= "Lee"
pw= "1234"
print(id == "Lee" and pw == "1234")
print(id == "Lee" or pw == "1234")
```

```python
# 예제 11. 논리 부정(not) 연산자를 사용하여 식의 결과를 확인해 보자.
a = 10
b = 20
print(not a > 5)
print(not b < 15)
```

### 🙌 활동 13

**1.** 키(m)와 몸무게(kg)를 변수에 저장하고 **체질량 지수**를 계산하는 프로그램을 완성해 보자.

**2.** 1번에서 구한 체질량 지수(bmi)가 **정상 범위**에 속하는지 그 결과를 `True`나 `False`로 출력하는 명령문을 완성해 보자.
(정상 범위: 체질량 지수가 **18.5 이상 23 미만**)

```python
# 활동 13 ✏️ 빈칸을 채워 보자.
h = 1.7   # 키
w = 60    # 몸무게
bmi = w / (h ** 2)      # 1번: 체질량 지수 계산
print(bmi)
print(18.5 <= bmi < 23)   # 2번: 정상 범위(18.5 이상 23 미만) 여부 출력
```

### 🙌 활동 14

[혈압 단계 구분] 표와 [실행 결과]를 참고하여 **고혈압 단계를 알려 주는 프로그램**을 완성해 보자.

```python
# 활동 14 ✏️ 빈칸을 채워 보자.
a, b = 135, 88   # a는 수축기 혈압, b는 이완기 혈압
print("수축기 혈압", a, "이완기 혈압", b)
print("정상 혈압:", a < 120 and b < 80)
print("주의 혈압:", 120 <= a <= 129 and b < 80)
print("고혈압 전 단계:", 130 <= a <= 139 or 80 <= b <= 89)
print("고혈압:", a >= 140 or b >= 90)
```
