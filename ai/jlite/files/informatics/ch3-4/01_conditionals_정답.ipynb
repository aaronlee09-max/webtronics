```python
# 예제 19. 어린이 보호 구역은 최고 통행 속도가 30km이다.
speed = int(input("속도: "))
if speed > 30:
    print("과속입니다.")
```

```python
# 활동 18 ✏️ 조건문을 완성해 보자.
height = float(input("키: "))
if height < 140:
    print("더블스핀 놀이기구 탑승 불가")
```

```python
# 예제 20. 일반 도로와 어린이 보호 구역에서 자동차가 달릴 수 있는 속도를 안내해 보자.
road = input("도로 종류(어린이 보호 구역이면 a): ")
if road == 'a':
    print("어린이 보호 구역은 시속 30km입니다.")
else:
    print("일반 도로는 시속 50km입니다.")
```

```python
# 활동 19 ✏️ 빈칸에 알맞은 조건문을 작성해 보자.
age = int(input("나이를 입력하세요."))
if age <= 18:
    print("입장권 가격은 40,000원입니다.")
else:
    print("입장권 가격은 50,000원입니다.")
```

```python
# 예제 21. 도로 종류를 입력하여 제한 속도를 알려 주는 프로그램을 작성해 보자.
road = input("도로 종류: ")
if road == "고속 도로":
    print("제한 속도는 100km/h입니다.")
else:
    child = input("어린이 보호 구역인가요?(y/n): ")
    if child == "y":
        print("제한 속도는 30km/h입니다.")
    else:
        print("제한 속도는 50km/h입니다.")
```

```python
# 예제 22. 시험 점수를 입력하여 해당하는 학점을 출력해 보자.
score = int(input("점수: "))
if score >= 80:
    print("A")
elif score >= 60:
    print("B")
else:
    print("C")
```

```python
# 활동 20 ✏️ 나머지 조건문을 완성해 보자.
height = int(input('키: '))
if height < 100:
    print('탑승 불가')
elif height < 130:
    print('보호자 동행 시 탑승 가능')
elif height < 195:
    print('탑승 가능')
else:
    print('탑승 불가')
```
