```python

    명령문 1
명령문 2
```

- 조건식이 참일 때만 명령문 1을 실행하고 명령문 2를 실행한다.
- 조건식의 결과가 거짓이면 명령문 1은 실행하지 않고 명령문 2를 실행한다.
- 명령문 2는 조건식의 결과와 상관없이 무조건 실행한다.

## 2️⃣ 중첩 조건문

조건문 안에 **또 다른 조건문을 포함**하는 구조이다.

```python
if 조건식 1 :
    if 조건식 2 :
        명령문 1
    else :
        명령문 2
else :
    명령문 3
```

- 조건식 1이 참일 때 조건식 2도 참이면 명령문 1을 실행한다.
- 조건식 1은 참이고 조건식 2가 거짓이면 명령문 2를 실행한다.
- 조건식 1이 거짓이면 명령문 3을 실행한다.

> 💡 if문 안에 또 다른 if문이 들어갈 수 있고, else문 안에 또 다른 if문이 들어갈 수도 있다.

## 3️⃣ 다중 조건문

선택 사항이 **여러 개**일 경우 사용하는 조건문으로 여러 개의 조건식에 따라 명령을 다르게 실행할 경우 사용한다.

```python
if 조건식 1 :
    명령문 1
elif 조건식 2 :
    명령문 2
else :
    명령문 3
```

- 조건식 1이 참이면 명령문 1을 실행하고 if문을 벗어난다.
- 조건식 1이 거짓이면 elif문의 조건식 2를 비교하여 그 결과가 참이면 명령문 2를 실행하고 if문을 벗어난다.
- 조건식 1과 조건식 2가 모두 거짓이면 명령문 3을 실행한다.

> 💡 elif문에는 반드시 조건식이 있어야 하며, else문 없이 여러 개의 elif문으로 다중 조건문을 만들 수 있다.
