같은 펭귄 데이터를 정규화한 뒤 로지스특 회귀와 k-NN 모델의 정확도 및 혼동 행렬을 비교해 보자.


## 단계 0: 준비 (라이브러리 설치)


```python
import importlib, sys, subprocess
packages=[('pandas','pandas'),('matplotlib','matplotlib'),('seaborn','seaborn'),('sklearn','scikit-learn')]
for module,pip_name in packages:
    try: importlib.import_module(module)
    except ModuleNotFoundError: subprocess.run([sys.executable,'-m','pip','install','-q',pip_name], check=True)

from sklearn import set_config
set_config(display="text")

print('라이브러리 준비 완료')
```

## 단계 1~2: 문제 정의와 데이터 전처리


```python
# penguins_size.csv를 불러오고 결측치 제거하기
import pandas as pd
df = pd.read_csv('penguins_size.csv')
df.dropna(inplace=True)
# 3개 특징 X와 species 타깃 y 선정하기
X = df[['culmen_length_mm', 'culmen_depth_mm', 'flipper_length_mm']]
y = df['species']
```

```python
# MinMaxScaler로 X를 0~1 범위로 정규화하기
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
X = scaler.fit_transform(X)
```

```python
# 훈련 70%, 테스트 30%로 계층 분할하기(random_state=42)
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)
```

## 단계 3: 모델 생성하기


```python
# LogisticRegression 모델(lr)과 k=5인 KNeighborsClassifier 모델(knn)을 학습하기
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
lr = LogisticRegression(max_iter=1000)
knn = KNeighborsClassifier(n_neighbors=5)
lr.fit(X_train, y_train)
knn.fit(X_train, y_train)
```

## 단계 4: 모델 평가하기


```python
# 두 모델의 테스트 정확도를 출력하기
print('로지스특 회귀:', lr.score(X_test, y_test))
print('kNN:', knn.score(X_test, y_test))
```

```python
# 로지스특 회귀 예측 결과의 혼동 행렬을 히트맵으로 나타내기
from sklearn.metrics import confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt
pred = lr.predict(X_test)
cm = confusion_matrix(y_test, pred)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=lr.classes_, yticklabels=lr.classes_)
plt.xlabel('예측')
plt.ylabel('실제')
plt.show()
```
