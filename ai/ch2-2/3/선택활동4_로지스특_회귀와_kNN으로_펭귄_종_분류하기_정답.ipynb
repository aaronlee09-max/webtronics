```python
import importlib, sys, subprocess
packages=[('pandas','pandas'),('matplotlib','matplotlib'),('seaborn','seaborn'),('sklearn','scikit-learn')]
for module,pip_name in packages:
    try: importlib.import_module(module)
    except ModuleNotFoundError: subprocess.run([sys.executable,'-m','pip','install','-q',pip_name], check=True)
from sklearn import set_config
set_config(display='text')
print('라이브러리 준비 완료')
```

```python
import pandas as pd
df = pd.read_csv('penguins_size.csv')
df.dropna(inplace=True)
X = df[['culmen_length_mm', 'culmen_depth_mm', 'flipper_length_mm']]
y = df['species']
```

```python
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
X = scaler.fit_transform(X)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
lr = LogisticRegression(max_iter=1000)
knn = KNeighborsClassifier(n_neighbors=5)
lr.fit(X_train, y_train)
knn.fit(X_train, y_train)
```

```python
print('로지스특 회귀:', lr.score(X_test, y_test))
print('kNN:', knn.score(X_test, y_test))
```

```python
from sklearn.metrics import confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt
pred = lr.predict(X_test)
cm = confusion_matrix(y_test, pred)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=lr.classes_, yticklabels=lr.classes_)
plt.show()
```
