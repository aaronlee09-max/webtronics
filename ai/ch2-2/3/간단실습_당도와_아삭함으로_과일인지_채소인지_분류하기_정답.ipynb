# 간단실습: 당도와 아삭함으로 과일인지 채소인지 분류하기

`food.csv`의 당도(`sweet`)와 아삭함(`crunchy`)을 이용하여 k-최근접 이웃 분류 모델을 구현해 보자.

```python
import os
os.environ["PYTHONUTF8"] = "1"

import sklearn
sklearn.set_config(display="text")  # HTML 대신 텍스트로 출력
import importlib, sys, subprocess
packages=[('pandas','pandas'),('matplotlib','matplotlib'),('seaborn','seaborn'),('sklearn','scikit-learn')]
for module,pip_name in packages:
    try: importlib.import_module(module)
    except ModuleNotFoundError: subprocess.run([sys.executable,'-m','pip','install','-q',pip_name], check=True)

from sklearn import set_config
set_config(display="text")

print('라이브러리 준비 완료')
```

## 단계 1: 문제 정의하기

**당도와 아삭함으로 과일인지 채소인지 분류해 보자.**


## 단계 2: 데이터 수집 및 전처리하기

`food.csv`를 불러오고 특징(X)과 타깃(y)을 선정한다.


```python
# pandas로 food.csv를 불러와 df에 저장하고 처음 5행 확인하기
import pandas as pd
df = pd.read_csv('food.csv')
df.head()
```

```python
# seaborn으로 x=sweet, y=crunchy, hue=class인 산점도 그리기
import seaborn as sns
sns.scatterplot(data=df, x='sweet', y='crunchy', hue='class')
```

```python
# sweet와 crunchy를 특징 X로, class를 타깃 y로 선정하기
X = df[['sweet','crunchy']].values
y = df['class']
y
```

## 단계 3: 모델 생성하기

**💡 새 API: `KNeighborsClassifier`**

- `KNeighborsClassifier(n_neighbors=3)`은 가장 가까운 이웃 3개의 다수결로 클래스를 정하는 모델을 만든다.
- `knn.fit(X, y)`는 특징 X와 정답 y의 관계를 모델에 학습시킨다.
- 사용 형식: `모델 = KNeighborsClassifier(n_neighbors=k)` → `모델.fit(특징, 타깃)`


```python
# KNeighborsClassifier를 가져와 k=3인 knn 모델을 만들고 X, y로 학습하기
from sklearn.neighbors import KNeighborsClassifier
knn = KNeighborsClassifier(n_neighbors=3)
knn.fit(X, y)
```

## 단계 4: 모델 평가하기

**💡 새 API: `predict()`**

- `모델.predict([[특징1, 특징2]])`는 새 데이터 한 개의 클래스를 예측한다.
- 모델은 여러 행을 입력받으므로 새 데이터 한 개도 `[[2, 5]]`처럼 2차원으로 넣는다.


```python
# 당도 2, 아삭함 5인 식품의 클래스를 예측하기
knn.predict([[2,5]])
```
