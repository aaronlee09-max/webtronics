14개 서비스 만족도 특징을 사용해 고객 만족 여부를 분류하고, 중요한 서비스 특징을 확인해 보자.


## 단계 0: 준비 (라이브러리 설치)


```python
import importlib, sys, subprocess
packages=[('pandas','pandas'),('matplotlib','matplotlib'),('seaborn','seaborn'),('sklearn','scikit-learn')]
for module,pip_name in packages:
    try: importlib.import_module(module)
    except ModuleNotFoundError: subprocess.run([sys.executable,'-m','pip','install','-q',pip_name], check=True)

from sklearn import set_config
set_config(display="text")

print('라이브러리 준비 완료')
```

## 단계 1: 문제 정의하기

**항공사는 고객 만족도를 높이기 위해 어떤 서비스에 중점을 두어야 할까?**


## 단계 2: 데이터 수집 및 전처리하기


```python
# airline_passenger_satisfaction.csv를 df로 불러오기
import pandas as pd
df = pd.read_csv('airline_passenger_satisfaction.csv')
# 9~22번 열을 특징 X로, 마지막 열을 타깃 y로 선정하기
X = df.iloc[:, 9:23]
y = df.iloc[:, -1]
print(X.columns.tolist())
print(y.name)
```

```python
# 훈련 70%, 테스트 30%로 계층 분할하기(random_state=42)
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)
```

## 단계 3: 모델 생성하기

선택 활동 4에서 배운 결정트리와 로지스특 회귀를 다시 사용한다. `max_depth=5`는 트리가 지나치게 깊어져 훈련 데이터를 외우는 현상을 줄이기 위한 제한이다.


```python
# DecisionTreeClassifier(max_depth=5)와 LogisticRegression(max_iter=1000)을 학습하기
from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LogisticRegression
dt = DecisionTreeClassifier(max_depth=5, random_state=42)
lr = LogisticRegression(max_iter=1000)
dt.fit(X_train, y_train)
lr.fit(X_train, y_train)
```

## 단계 4: 모델 평가하기


```python
# 두 모델의 테스트 정확도를 비교하기
print('결정트리:', dt.score(X_test, y_test))
print('로지스특 회귀:', lr.score(X_test, y_test))
```

```python
# 결정트리의 특징 중요도를 내림차순으로 출력하고 상위 5개 시각화하기
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
importance = pd.Series(dt.feature_importances_, index=X.columns).sort_values(ascending=False)
print(importance)
sns.barplot(x=importance.head(5).values, y=importance.head(5).index)
plt.show()
```
