## 단계 2: 데이터 수집 및 전처리하기


```python
# pandas로 penguins_size.csv를 불러와 df에 저장하고 정보 확인하기
import pandas as pd
df = pd.read_csv('penguins_size.csv')
df.info()
```

```python
# 열별 결측치 개수를 확인한 뒤 결측치가 있는 행 삭제하기
print(df.isnull().sum())
df.dropna(axis=0, inplace=True)
df.info()
```

```python
# 부리 길이, 부리 깊이, 날개 길이를 X로, species를 y로 선정하기
X = df[['culmen_length_mm', 'culmen_depth_mm', 'flipper_length_mm']]
y = df['species']
y.unique()
```

```python
# 데이터를 훈련 70%, 테스트 30%로 계층 분할하기(random_state=42)
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y)
print(X_train.shape, X_test.shape, y_train.shape, y_test.shape)
```

## 단계 3: 모델 생성하기

**💡 새 API: `DecisionTreeClassifier`**

- `DecisionTreeClassifier()`는 특징값에 대한 질문을 이어 붙이는 분류 모델을 만든다.
- `dt.fit(X_train, y_train)`은 훈련 데이터로 분할 기준을 학습한다.
- `dt.score(X, y)`는 맞힌 비율인 정확도를 반환한다.


```python
# DecisionTreeClassifier 모델(dt)을 만들고 훈련 데이터로 학습하기(random_state=42)
from sklearn.tree import DecisionTreeClassifier
dt = DecisionTreeClassifier(random_state=42)
dt.fit(X_train, y_train)
```

```python
# plot_tree로 최대 깊이 2까지 결정트리 시각화하기
from sklearn.tree import plot_tree
import matplotlib.pyplot as plt
plt.figure(figsize=(12, 6))
plot_tree(dt, max_depth=2, feature_names=list(X.columns), class_names=list(dt.classes_), filled=True)
plt.show()
```

## 단계 4: 모델 평가하기


```python
# 테스트 데이터를 예측하고 정확도 구하기
from sklearn.metrics import accuracy_score
pred = dt.predict(X_test)
print(accuracy_score(y_test, pred))
```
