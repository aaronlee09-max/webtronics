```python
# 판다스 라이브러리 가져오기
import pandas as pd
# 데이터 프레임(df) 생성하기
df = pd.read_csv('penguins_size.csv')
# 데이터 프레임 5행 출력하기
df.head()
```

```python
# 데이터 속성 확인하기
df.info()
```

```python
# 데이터 통계 정보 요약
df.describe()
```

```python
# 막대그래프(countplot) 그리기
import seaborn as sns
sns.countplot(x=df['species'])
```

```python
# 열별 결측치 개수 확인하기
df.isnull().sum()
```

```python
# 결측치가 있는 행 삭제하기
df.dropna(axis=0, inplace=True)
print(len(df))
```

```python
# 부리 길이 상자그림 그리기
import matplotlib.pyplot as plt
plt.figure(figsize=(6, 2))
sns.boxplot(data=df, x='culmen_length_mm')
plt.show()
```

```python
#종의 고유한 값 종류 출력
df['species'].unique()
```

```python
#성별의 고유한 값 종류 출력
df['sex'].unique()
```

```python
# 잘못된 성별 값이 있는 행 찾기
df[df['sex'] == '.']
```

```python
# 잘못된 행 삭제하기
df = df[df['sex'] != '.']
print(len(df))
```

```python
# MinMaxScaler 생성하기
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
# 수치형 속성값을 최소-최대 정규화 처리
num = ['culmen_length_mm', 'culmen_depth_mm', 'flipper_length_mm', 'body_mass_g']
df[num] = scaler.fit_transform(df[num])
df[num].head()
```

```python
# 부리 길이와 부리 깊이 산점도 그리기
sns.scatterplot(data=df, x='culmen_length_mm', y='culmen_depth_mm', hue='species')
```

```python
# 부리 깊이와 날개 길이 산점도 그리기
sns.scatterplot(data=df, x='culmen_depth_mm', y='flipper_length_mm', hue='species')
```

```python
# 부리 길이와 체질량 산점도·회귀선 그리기
sns.regplot(data=df, x='culmen_length_mm', y='body_mass_g')
```

```python
# 수치형 속성 상관계수 히트맵 그리기
num = ['culmen_length_mm', 'culmen_depth_mm', 'flipper_length_mm', 'body_mass_g']
sns.heatmap(df[num].corr(), annot=True)
```
