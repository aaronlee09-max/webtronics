# 선택 활동: 우리나라의 지진 발생 현황을 분석해 보기


```python
# CSV 자동 준비 (Colab · JupyterLite · 로컬 공통)
# 파일이 없으면 GitHub raw에서 받아 cwd에 저장 → pd.read_csv("파일명") 그대로 동작
import sys
from pathlib import Path

def ensure_csv(filename, repo_path):
    candidates = [
        Path(filename),
        Path(repo_path),
        Path("files") / repo_path,
        Path("/files") / repo_path,
        Path(repo_path).name,
    ]
    for c in candidates:
        try:
            if c.is_file():
                if not Path(filename).exists():
                    Path(filename).write_bytes(c.read_bytes())
                return filename
        except Exception:
            pass
    url = f"https://raw.githubusercontent.com/aaronlee09-max/informatics/main/{repo_path}"
    try:
        if sys.platform == "emscripten":
            from pyodide.http import open_url
            Path(filename).write_text(open_url(url).read(), encoding="utf-8")
        else:
            import urllib.request
            urllib.request.urlretrieve(url, filename)
        print("CSV 준비:", filename)
        return filename
    except Exception as e:
        print("CSV 준비 실패 → URL 직접 사용:", url, e)
        return url

ensure_csv('earthquake.csv', 'ai/ch2-1/earthquake.csv')
ensure_csv('penguins_size.csv', 'ai/ch2-1/penguins_size.csv')

import importlib, sys, subprocess
packages = [('pandas','pandas'),('seaborn','seaborn'),('matplotlib','matplotlib'),('koreanize_matplotlib','koreanize-matplotlib')]
for module_name, pip_name in packages:
    try:
        importlib.import_module(module_name)
    except ModuleNotFoundError:
        subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', pip_name])
print('라이브러리 준비 완료')
```

```python
import pandas as pd
df = pd.read_csv('earthquake.csv', encoding='utf-8')
df.head()
```

```python
df.isnull().sum()
```

```python
df.dropna(axis=0, inplace=True)
```

```python
df['year'] = pd.to_datetime(df['시간']).dt.year
df['year'].value_counts().sort_index()
```

```python
import seaborn as sns
sns.scatterplot(data=df, x='year', y='규모')
```

```python
df['지역']=df['위치'].str[:2]
df.head()
```

```python
import matplotlib.pyplot as plt
counts = df['지역'].value_counts()
plt.figure(figsize=(8, 8))
plt.pie(counts, labels=counts.index, autopct='%.1f%%')
plt.show()
```
