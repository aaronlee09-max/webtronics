# 롤러코스터 파헤치기 정답


```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('rollercoasters.csv')
```

```python
df.info()
print(df.shape)
df.head()
df.describe()
```

```python
df2 = df.sort_values(by='max_speed', ascending=False)
df2.iloc[0:10, [1, 2, 4, 10]]
```

```python
x1 = df['rollercoaster_type'].unique()
x2 = df['rollercoaster_type'].nunique()
rtc = df['rollercoaster_type'].value_counts()
print(x1, x2)
rtc
```

```python
plt.figure(figsize=(10, 6))
plt.bar(rtc.index, rtc.values)
plt.xticks(rotation=90)
plt.show()
```

```python
corr = df.corr(numeric_only=True)
corr2 = corr['excitement']
plt.xticks(rotation=90)
plt.bar(corr2.index, corr2.values)
plt.show()
```
