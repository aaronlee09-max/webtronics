# 정답 노트북


```
score = int(input('점수: '))
if score >= 90:
    print('A')
elif score >= 80:
    print('B')
elif score >= 70:
    print('C')
else:
    print('F')
```

```
n = int(input('숫자: '))
if n % 2 == 0:
    print('짝수')
else:
    print('홀수')
```

```
age = int(input('나이: '))
if age >= 18:
    print('대인')
else:
    print('소인')
```
