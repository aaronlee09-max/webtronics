# 정답 노트북


```
for i in range(5):
    print(i)
```

```
d = int(input('몇 단? '))
for i in range(1, 10):
    a = d * i
    print(f'{d} * {i} = {a}')
```

```
for i in range(2):
    for j in range(3):
        print(f'i: {i}, j: {j}')
```

```
for d in range(2, 10):
    for i in range(1, 10):
        a = d * i
        print(f'{d} * {i} = {a}')
```

```
popcorn_options = ['기본', '카라멜', '어니언']
drink_options = ['생수', '탄산음료']
for popcorn in popcorn_options:
    for drink in drink_options:
        print(f'세트 메뉴: 팝콘 - {popcorn}, 음료 - {drink}')
```

```
a = 1
while a <= 5:
    print(a)
    a += 1
```

```
for a in [1, 2, 3, 4, 5]:
    print(a)
```

```
while True:
    user_input = input('게임을 계속하려면 아무 키나 누르세요. 종료하려면 q: ')
    if user_input.lower() == 'q':
        print('프로그램 나가기')
        break
```

```
total = 0
for i in range(1, 101):
    if i % 5 == 0:
        total += i
print(total)
```

```
import random
num = random.randint(1, 10)
guess = int(input('1부터 10 사이 숫자: '))
for i in range(10):
    if guess == num:
        print('정답입니다!')
        break
    elif guess < num:
        print('UP')
    else:
        print('DOWN')
    guess = int(input('다시 입력: '))
```
