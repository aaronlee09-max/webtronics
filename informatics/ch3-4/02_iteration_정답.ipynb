```python
import time
for i in range(5, 0, -1):
    print(i, "초 남았습니다.")
    time.sleep(1)
print("타이머 종료!")
```

```python
d = int(input("몇 단? "))
for i in range(1, 10):
    a = d * i
    print(f"{d} * {i} = {a}")
```

```python
for i in range(2):
    for j in range(3):
        print(f"i: {i}, j: {j}")
```

```python
for d in range(2, 10):
    for i in range(1, 10):
        a = d * i
        print(f"{d} * {i} = {a}")
```

```python
popcorn_options = ["기본", "카라멜", "어니언"]
drink_options = ["생수", "탄산음료"]
for popcorn in popcorn_options:
    for drink in drink_options:
        print(f"세트 메뉴: 팝콘 - {popcorn}, 음료 - {drink}")
```

```python
while True:
    user_input = input("종료하려면 q 또는 Q: ")
    if user_input.lower() == 'q':
        print("프로그램 나가기")
        break
```

```python
total = 0
for i in range(1, 101):
    if i % 5 == 0:
        total += i
print(total)
```

```python
import random
num = random.randint(1, 10)
guess = int(input("1부터 10 사이 숫자: "))
for i in range(10):
    if guess == num:
        print("정답입니다!")
        break
    elif guess < num:
        print("UP!")
    else:
        print("DOWN!")
    guess = int(input("다시 입력: "))
```
