# 정답 노트북


```
a = 'HAPPY BIRTHDAY'
b = '2U'
print(a[1])
print(a[6:11])
print(a[0:5])
print(a[11:])
print(a + b)
print(b * 2)
```

```
a = 'HAPPY BIRTHDAY!'
print(a[0] + a[6] + a[11])
```

```
a = 25
b = 12
print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a % b)
print(a // b)
print(a ** b)
```

```
n = 10
print(n == 5)
print(n != 5)
print(n > 5)
print(n <= 5)
print(5 <= n < 10)
```

```
id= 'Lee'
pw= '1234'
print(id == 'Lee' and pw == '1234')
print(id == 'Lee' or pw == '1234')
```

```
a = 10
b = 20
print(not a > 5)
print(not b < 15)
```

```
h = 1.7
w = 60
bmi = w / (h ** 2)
print(bmi)
print(18.5 <= bmi < 23)
```

```
a, b = 135, 88
print('수축기 혈압', a, '이완기 혈압', b)
print('정상 혈압:', a < 120 and b < 80)
print('주의 혈압:', 120 <= a < 140 or 80 <= b < 90)
print('고혈압 전 단계:', 140 <= a < 160 or 90 <= b < 100)
print('고혈압:', a >= 160 or b >= 100)
```
