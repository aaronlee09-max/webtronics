## ④ 표준입출력

컴퓨터 프로그램은 사용자로부터 데이터를 **입력**받아 **처리**하고 결과를 사용자에게 **출력**하는 단계로 진행된다.

### ⑴ input( )

키보드로 데이터를 입력받는 명령어로 **입력받은 값은 문자열로 처리**된다.

> **형식** `변수명 = input( )`

**형 변환** : `input( )으로 입력한 값은 모두 문자열로 처리되기 때문에, `int( )`나 `float( )을 이용하여 자료형을 변환해야 한다.

- `int("2025")` → 정숫값 2025
- `float("3.8")` → 실숫값 3.8
- `int("3.8")` → ❌ 오류 (소수점을 포함하는 문자열은 정수형으로 변환할 수 없다.)

````python
# 예제 12. 키보드로 데이터를 입력받아 출력해 보자.
# input( )으로 데이터를 입력받을 때, 지시어를 포함할 수 있다.
name = input("이름: ")
print(name)
````

### ⑵ print( )

모니터 화면에 데이터를 출력하는 명령어로 형식을 지정하여 원하는 형태로 데이터를 출력할 수 있다.

> **형식**
> ```python
> print(값)                # 값 출력하기
> print(변수명)            # 변수에 저장된 값 출력하기
> print(수식)              # 수식 결과 출력하기
> print(값1, 값2, 값3, ...) # 여러 개의 값 한 줄에 출력하기
> ```

````python
# 예제 13. 표준 출력 명령어를 이용하여 다양한 형태로 출력해 보자.
print(10)
a = 20
print(a)
print(3 + 5)
print("합", 3, 5)
````

### 🙌 활동 16

놀이공원 [입장료 구분] 표와 [실행 결과]를 참고하여 **놀이공원 입장료를 출력하는 프로그램**을 완성해 보자.

| 구분 | 입장료(원) |
|------|-----------|
| 대인(만 18세 이상) | 50,000 |
| 소인(만 18세 미만) | 30,000 |

```
실행 결과

소인: 2
160000
```

> 💡 `int(input( ))`으로 입력값을 정수형으로 바꿀 수 있어요.

````python
# 활동 16 ✏️ 빈칸을 채워 보자.
adult = 50000
child = 30000
n = int(input("대인: "))
m = int(input("소인: "))
print(adult * n + child * m)
````
