```python
import nbformat

def count_markdown_above(filepath, ignore_spaces=False):
    """
    현재 노트북 파일을 읽어, 이 코드가 작성된 셀 바로 위의 마크다운 셀 글자 수를 세는 함수
    """
    with open(filepath, 'r', encoding='utf-8') as f:
        nb = nbformat.read(f, as_version=4)
    
    # 이 코드가 작성된 셀 자체를 찾기 위한 고유 식별 문자열
    my_signature = "def count_markdown_above(filepath"
    
    for i, cell in enumerate(nb.cells):
        # 노트북 데이터에서 현재 코드 셀의 위치를 찾음
        if cell.cell_type == 'code' and my_signature in cell.source:
            # 현재 셀의 바로 위(i-1) 셀이 마크다운인지 확인
            if i > 0 and nb.cells[i-1].cell_type == 'markdown':
                target_text = nb.cells[i-1].source
                
                # 공백 및 줄바꿈 제거 옵션 적용
                if ignore_spaces:
                    target_text = target_text.replace(" ", "").replace("\n", "").replace("\r", "")
                
                # 텍스트가 맞는지 확인할 수 있도록 원본 텍스트 앞부분 30자 추출
                preview = nb.cells[i-1].source[:30].replace("\n", " ")
                return len(target_text), f"{preview}..."
            else:
                return 0, "오류: 바로 위 셀이 마크다운 셀이 아니거나 비어있습니다."
                
    return 0, "오류: 파일 내에서 코드를 찾을 수 없습니다. 파일명을 확인해 주세요."

# --- 실행 부분 ---
# ★주의: 실행 전 반드시 노트북을 저장(Ctrl+S)해야 최신 상태의 텍스트를 읽어옵니다.

notebook_filename = 'book.ipynb' # 실제 파일명으로 수정하세요. (예: 'test.ipynb')

count, message = count_markdown_above(notebook_filename, ignore_spaces=True)
print(f"글자 수(공백 제외): {count}자")
```
